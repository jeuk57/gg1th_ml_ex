#### 경력과 연봉의 관계 분석을 통해, 누군가 입사했을 때, 그사람의 경력에 맞는 연봉을 제시해 줄 수 있도록 합니다.

## 데이터 준비

```python
df = pd.read_csv('datas_ml/Salary_Data.csv')
df.head()
```

```python
# 결측치 확인
df.info()
```

```python
df.isna().sum()
```

```python
# 우리가 예측하려고 하는 것은? 연봉
```

## x, y 데이터 분할

```python
y = df['Salary']
```

```python
x = df['YearsExperience']
```

```python
y.shape
```

```python
x.shape
```

```python
# x는 무조건 2차원이어야 함.
# x.to_frame()
x = x.to_frame()
x.head()
```

## 피처스케일링
- 독립변수 1개 이므로, 따로 해줄 필요 없다.
### 스케일링이 필요할 때
- 피처의 스케일 차이가 너무 크면
- 수치 불안정 → 계산 느림 → 가중치 해석 어려움

## 학습 데이터, 테스트 데이터 분할

```python
from sklearn.model_selection import train_test_split
```

```python
# 학습용과 테스트용으로 나눔
train_test_split(x, y, test_size=0.2, random_state=7)
```

```python
# 대입 변수 순서가 중요함.
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=7
)
```

```python
x_train.shape, y_train.shape
```

```python
x_test.shape, y_test.shape
```

## 모델 학습하기

```python
# 연봉 예측 모델 만들기 => 예측 결과 숫자, 실수 숫자가 나오는 인공지능 만들기  => 회귀 regression
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# 깡통 AI
lr_model = LinearRegression()
```

```python
# 학습 되기 전의 기울기
lr_model.coef_
```

```python
# 학습 되기 전의 절편
lr_model.intercept_
```

```python
# 회귀 모델 객체
lr_model
```

### 학습하기
- 문제(x)와 정답(y)를 넣어 학습하기

```python
# 학습 데이터를 넣어서 학습
lr_model.fit(x_train, y_train)
```

```python
# y = ax + b
# a, b를 찾는 거
# 기울기
lr_model.coef_
```

```python
# 절편
lr_model.intercept_
```

## 모델 테스트

```python
# 테스트 데이터를 넣어, 결과 예측하기
y_pred = lr_model.predict(x_test)
y_pred
```

## 모델 평가
- 모델이 테스트한 결과(y_pred)와 실제값(y_test)의 오차의 제곱, 합의 평균 계산

```python
# 정답
y_test
```

```python
# 예측값
y_pred
```

```python
x_test
```

```python
# 오차 제곱
(y_test - y_pred)**2
```

```python
# 오차 : error의 제곱을 합한 후 데이터의 갯수로 나눔<- 오차제곱을 구한후 평균 구함 (MSE)
((y_test - y_pred)**2).mean() 
```

```python
# RMSE (MSE는 너무크기 때문에 RMS구함.)
np.sqrt(((y_test - y_pred)**2).mean())
```

```python
# MAE (절대값으로)
abs(y_test - y_pred).mean()
```

## 오차를 시각적으로 확인하기

```python
df_test = y_test.to_frame()
```

```python
df_test['y_pred'] = y_pred
```

```python
df_test
```
