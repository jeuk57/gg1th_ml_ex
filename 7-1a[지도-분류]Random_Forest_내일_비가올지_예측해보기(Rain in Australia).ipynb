## 결측치 채우기
### 수치형 컬럼 결측처리
- numerical value를 가진 column은 중위값(meidan)으로 채우기

```python
# Sunshine 열의 결측치를 이상치의 영향을 덜 받는 중앙값으로 채움
#df['Sunshine'].fillna(df['Sunshine'].median())
df['Sunshine'] = df['Sunshine'].fillna(df['Sunshine'].median())
```

```python
# 모든 숫자형 열을 반복하며 결측치가 있는 열은 중앙값으로 채움
for col in num_cols:
    if df[col].isnull().mean() > 0:
        col_median = df[col].median()
        df[col] = df[col].fillna(col_median)
```

### 범주형 컬럼 결측처리 
- null이 있는 범주형 컬럼 확인

```python
# 결측치가 남은 범주형 열과 결측치 비율을 출력
for col in cat_cols:
    if df[col].isnull().mean() > 0:
        print(col, (df[col].isnull().mean()))
```

```python
# WindGustDir에서 가장 자주 나오는 방향(최빈값) 하나를 확인
# df['WindGustDir'].mode()
df['WindGustDir'].mode()[0]
```

```python
# 방법1
# categorical value를 가진 column은 최빈값(mode)으로 결측치 채우기
# RainTomorrow 는 target 이기 때문에 결측치 처리하지 않음.(일반적인 선택)
df['WindGustDir'] = df['WindGustDir'].fillna(df['WindGustDir'].mode()[0])
df['WindDir9am'] = df['WindDir9am'].fillna(df['WindDir9am'].mode()[0])
df['WindDir3pm'] = df['WindDir3pm'].fillna(df['WindDir3pm'].mode()[0])
df['RainToday'] = df['RainToday'].fillna(df['RainToday'].mode()[0])
```

## target의 null 값처리
- RainTomorrow는 target이기 때문에 nan은 drop으로 처리
- RainTomorrow 예측 값이므로 null 값 삭제함

```python
# 현재 데이터의 행과 열 개수를 확인
df.shape
```

```python
# 타겟 결측치 제거 전의 데이터 크기를 한 번 더 확인
df.shape
```

```python
# 타겟인 RainTomorrow에 결측치가 남아 있는지 확인
df.isnull().sum()
```

```python
# 타겟 결측치를 제거한 데이터의 앞 5행을 확인
df.head()
```

# Yes/No 값에 대한 변환

```python
# 모델이 학습할 수 있도록 비 여부 Yes/No를 1/0으로 변환
df['RainToday'] = df['RainToday'].map({'No': 0, 'Yes': 1})
df['RainTomorrow'] = df['RainTomorrow'].map({'No': 0, 'Yes': 1})

print(df['RainToday'].dtype, df['RainTomorrow'].dtype)
```

## categorical value에 one-hot encoding 적용하기

```python
# 앞에서 선택한 범주형 열 목록을 확인
cat_cols
```

```python
# 지역과 바람 방향 문자열을 모델이 학습할 수 있는 원-핫 인코딩 열로 변환
df = pd.get_dummies(df, columns = ['Location', 'WindGustDir', 'WindDir9am', 'WindDir3pm'])
```

```python
# 범주형 열이 True/False 인코딩 열로 변환되었는지 확인
df.head()
```

```python
# 원-핫 인코딩으로 열이 늘어난 후의 데이터 크기를 확인
df.shape
```

# ML을 위한 학습, 테스트 데이터 준비

```python
# [문제] 특징 변수와 타겟 변수 분할 하기
# RainTomorrow를 제외한 모든 열을 예측에 사용할 특징(X)로 설정
X = df.drop('RainTomorrow', axis=1)
# 내일 비 여부를 예측할 타겟(y)으로 설정
y = df['RainTomorrow']
```

```python
# 모델에 넣을 특징 데이터의 앞 5행을 확인
X.head()
```

```python
# 타겟 데이터의 처음 5개 값을 확인
y[:5]
```

```python
# 비가 오지 않음(0)과 옴(1)의 개수를 비교해 클래스 불균형을 확인
y.value_counts()
```

```python
# [문제]  학습, 테스트 데이터 분할 하기
# 학습 : 테스트 = 7 : 3 
# 불균형 데이터 이므로 stratify = y 로 설정
# 재현을 위해 seed 값은 2026 설정
# stratify=y는 학습용과 테스트용 데이터의 타겟 비율을 원본과 비슷하게 유지함
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=2026
)
```

```python
# 분할된 학습용·테스트용 특징과 타겟의 크기를 확인
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

# ML 모델 학습

## 결정 트리(Decision Tree)로 내일 비가 올지 안올지 예측해보기

```python
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
# [문제] 결정 트리 모델 학습 및 예측 수행
# 결정 트리 분류 모델을 생성하고 실행할 때마다 같은 결과가 나오도록 시드를 고정
dt_clf = DecisionTreeClassifier(random_state=2026)
# 학습 데이터의 특징과 정답으로 모델을 학습
dt_clf.fit(X_train, y_train)

# 학습에 사용하지 않은 테스트 데이터의 클래스를 예측
y_pred = dt_clf.predict(X_test)
```

```python
# 결정 트리가 예측한 0/1 결과를 확인
y_pred
```

```python
# [문제] 결정 트리 정확도 출력
# 전체 테스트 데이터 중 올바르게 예측한 비율을 계산
score = accuracy_score(y_test, y_pred)
print('결정 트리(Decision Tree) Accuracy :',score)
```

### [문제] 정확도, 정밀도, 재현율, f1-스코아를 값을 확인해 보세요.

```python
from sklearn.metrics import classification_report
# [문제] classification_report() 결과 출력
# 클래스별 정밀도, 재현율, F1 점수와 정확도를 한번에 확인
print(classification_report(y_test, y_pred))
```

```python
# [문제] roc_auc_score 출력, 소숫점 첫째자리까지만 출력
from sklearn.metrics import roc_auc_score
# 0과 1 클래스를 구분하는 성능을 계산하고 소수점 첫째 자리로 표시
roc_auc = roc_auc_score(y_test, y_pred)
print(round(roc_auc, 1))
```

## 랜덤 포레스트(Random Forest)로  
- 참고 : https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html
- DT default = 100개를 만들어서 학습을 시킴

```python
from sklearn.ensemble import RandomForestClassifier
# [문제] RandomForestClassifier 분류 모델 학습 및 예측 수행
# 여러 결정 트리의 예측을 종합하는 랜덤 포레스트 모델을 생성
rf_clf = RandomForestClassifier(random_state=2026, n_jobs=-1)
# n_jobs=-1은 사용 가능한 CPU 코어를 모두 사용해 학습 속도를 높임
rf_clf.fit(X_train, y_train)

# 학습된 랜덤 포레스트로 테스트 데이터를 예측
y_pred = rf_clf.predict(X_test)
```

```python
# 랜덤 포레스트가 예측한 0/1 결과를 확인
y_pred
```

```python
# 정확도 평가
# 실제 정답과 예측값을 비교해 정확도를 계산
score = accuracy_score(y_test, y_pred)
print('랜덤 포레스트(Random Forest) Accuracy :', score)
```

```python
from sklearn.metrics import classification_report
# classification_report() 결과 출력
# 불균형 데이터이므로 정확도 외에 정밀도와 재현율도 함께 확인
print(classification_report(y_test, y_pred))
```

```python
# [문제] roc_auc_score 출력, 소숫점 둘째자리까지만 출력
from sklearn.metrics import roc_auc_score
# 랜덤 포레스트의 ROC-AUC를 계산하고 소수점 둘째 자리로 표시
roc_auc = roc_auc_score(y_test, y_pred)
print(round(roc_auc, 2))
```

# XGBoost 분류기

```python
from xgboost import XGBClassifier
# [문제] xgboost 학습 수행 및 예측
# 앞선 트리의 오류를 다음 트리가 보완하는 XGBoost 분류 모델을 생성
xgb_clf = XGBClassifier(random_state=2026)
# 학습 데이터로 XGBoost 모델을 학습
xgb_clf.fit(X_train, y_train)

# 학습된 XGBoost 모델로 테스트 데이터를 예측
y_pred = xgb_clf.predict(X_test)
```

```python
from sklearn.metrics import classification_report
# [문제] classification_report() 결과 출력
# XGBoost의 클래스별 정밀도, 재현율, F1 점수를 출력
print(classification_report(y_test, y_pred))
```

```python
# [문제] roc_auc_score 출력, 소숫점 둘째자리까지만 출력
from sklearn.metrics import roc_auc_score
# XGBoost의 ROC-AUC를 계산하고 소수점 둘째 자리로 표시
roc_auc = roc_auc_score(y_test, y_pred)
print(round(roc_auc, 2))
```
