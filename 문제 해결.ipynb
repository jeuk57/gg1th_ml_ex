### 데이터 설명
1. ph: pH 1. 물(0~14).량
2. Hardness비누를 침전시키는 물의 용량(mg/L)L.
3. Solid총 용해 고형물(ppm)pm.
4. Chloramin클로라민의 양(ppm).ppm.
5. Sulfmg/L 단위로 용해된 황산염의 양.mg/L.
6. Conducti물의 전기 전도도(단위: μS/cm).μS/cm.
7. Organic_c유기 탄소의 양(ppm).in ppm.
8. Trihalomeμg/L 단위의 트리할로메탄 양.in μg/L.
9. Tu NTU 단위로 물의 발광 특성을 측정함.r in NTU.
10. Po물이 인간이 섭취하기에 안전한지 여부를 나타냄
- 마실 수 있는 -> 1
- 마실 수 없는 -0potable -> 0

### 데이터 분포, 관계 시각화

### 4. Data Preprocess
- 각 컬럼별로 결측 데이터의 갯수가 몇개인지 확인하기

### nan이 하나라도 있는 index(행) 삭제

```python
# 결측치가 하나라도 있는 행을 제거하고 인덱스를 0부터 다시 정리
drop_row = water_df.dropna().reset_index(drop=True)
```

### 5. Data Split
- drop_row 데이터 프레임에서 독립변수 X 와 종속변수 y로 나누기

```python
# Potability는 정답이므로 제외하고, 나머지 9개 피처를 독립변수 X로 설정
X = drop_row.drop(columns='Potability')
```

```python
X.shape
```

```python
# 음용 가능 여부를 나타내는 Potability 열을 종속변수 y로 설정
y = drop_row['Potability']
```

```python
y.shape
```

```python
drop_row.info()
```

### 6. 피처 스케일링
- 데이터의 특성을 파악하여 피처 스케일 방법을 선택하고 독립변수에 적용하기
- 이상치가 많다고 판단하여 RobustScaler 사용

```python
from sklearn.preprocessing import RobustScaler

# 이상치의 영향을 적게 받는 RobustScaler 객체를 생성
robust_scaler = RobustScaler()
# X의 중앙값과 IQR을 학습하면서 스케일링까지 한번에 수행
X_scaled = robust_scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
x_cols = X.columns
x_cols
```

```python
X_scaled_df = pd.DataFrame(X_scaled, columns=x_cols)
X_scaled_df.head(3)
```

### 7. 데이터를 Train, Test로 나누기

```python
# 'Potability' 컬럼 값 비율 확인
sns.countplot(x=y)
```

- y 값을 균일하게 분할
- test data size 0.2
- 실행할 때마다 분할이 재현되도록 random_state 2024

```python
from sklearn.model_selection import train_test_split

# Potability의 0/1 비율을 유지하며 학습용 80%, 테스트용 20%로 분할
train_X, test_X, train_y, test_y = train_test_split(
    X_scaled_df, y, test_size=0.2, stratify=y, random_state=2024
)
```

```python
train_X.shape, train_y.shape
```

```python
test_X.shape, test_y.shape
```

### 8. KNN ML 모델 만들기 및 평가

```python
from sklearn.neighbors import KNeighborsClassifier
```

### GridSearchCV활용 Best Hyper Parameter 찾기

- 다음의 파라미터에 대해서 KNN 모델의 베스트 파라미터를 찾아보세요.

- n_neighbors : [2, 4, 6, 8, 10],
- p : [1, 2]
- cv : 3

```python
from sklearn.model_selection import GridSearchCV
```

```python
# KNN에서 비교할 이웃 수와 거리 계산 방식을 정의
# p=1은 맨하탄 거리, p=2는 유클리드 거리를 의미
params = {
    'n_neighbors': [2, 4, 6, 8, 10],
    'p': [1, 2]
}
```

```python
params
```

```python
# KNN 모델과 파라미터 조합을 3겹 교차검증으로 비교
knn_clf = KNeighborsClassifier()
grid_cv = GridSearchCV(
    estimator=knn_clf, param_grid=params, cv=3, scoring='accuracy', n_jobs=-1
)
```

```python
# 학습 데이터로 모델별 교차검증을 수행하고 최적 모델을 선택
grid_cv.fit(train_X, train_y)
```

### 모델 평가

```python
print(f"Best score of paramter search is: {grid_cv.best_score_:.4f}")
```

```python
grid_cv.best_params_
```

```python
print("Best parameter of best score is")
print(f"\t n_neighbors: {grid_cv.best_params_['n_neighbors']}")
print(f"\t p: {grid_cv.best_params_['p']}")
```

### 예측 및 평가

```python
# GridSearchCV가 찾은 최적 KNN 모델로 테스트 데이터를 예측
test_pred = grid_cv.predict(test_X)
```

```python
from sklearn.metrics import accuracy_score
# 실제 정답과 KNN 예측값을 비교해 정확도를 계산
test_acc = accuracy_score(test_y, test_pred)
print(f"test accuracy is {test_acc:.4f}")
```

```python
# 분류 성능 레포트: 클래스별 정밀도, 재현율, F1 점수를 확인
from sklearn.metrics import classification_report
print(classification_report(test_y, test_pred))
```

### 9. xgboost 모델 만들기 및 평가

```python
from xgboost import XGBClassifier

# XGBoost 분류 모델을 생성하고 학습 데이터로 학습
xgb_clf = XGBClassifier(random_state=2024)
xgb_clf.fit(train_X, train_y)
```

```python
train_pred = xgb_clf.predict(train_X)
test_pred = xgb_clf.predict(test_X)

train_acc = accuracy_score(train_y, train_pred)
test_acc = accuracy_score(test_y, test_pred)

print(f"train accuracy is {train_acc:.4f}")
print(f"test accuracy is {test_acc:.4f}")
```

```python
# classification_report() 결과 출력
print(classification_report(test_y, test_pred))
```

### 10. ML 모델의 feature importance 시각화

```python
#xgboost의 강력한 기능 Feature Importance 확인
from xgboost import plot_importance
from matplotlib import pyplot

# 학습된 XGBoost가 예측에 중요하게 사용한 피처를 시각화
plot_importance(xgb_clf)
pyplot.show()
```

### 11. 최적의 하이퍼 파라미터 찾기
- xgboost 모델이 과적합인 것 같다.
- xgboost 활용해 베스트 하이퍼 파리터 찾기
- GridSearchCV를 활용해 베스트 하이퍼 파라미터를 찾는 과정을 기술하고 모델을 평가해 보세요.

```python
# XGBoost에서 비교할 주요 하이퍼파라미터 조합을 정의
xgb_params = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'gamma': [0, 1, 2]
}

# 3겹 교차검증으로 정확도가 가장 높은 XGBoost 파라미터를 탐색
xgb_grid_cv = GridSearchCV(
    estimator=XGBClassifier(random_state=2024),
    param_grid=xgb_params,
    scoring='accuracy',
    cv=3,
    n_jobs=-1
)
xgb_grid_cv.fit(train_X, train_y)
```

```python
# 최고 교차검증 점수와 최적 파라미터를 확인
print(f'Best CV accuracy: {xgb_grid_cv.best_score_:.4f}')
print(f'Best parameters: {xgb_grid_cv.best_params_}')

# 최적 XGBoost 모델로 테스트 데이터를 예측하고 성능을 평가
xgb_best_pred = xgb_grid_cv.predict(test_X)
xgb_best_acc = accuracy_score(test_y, xgb_best_pred)
print(f'Best XGBoost test accuracy: {xgb_best_acc:.4f}')
print(classification_report(test_y, xgb_best_pred))
```
